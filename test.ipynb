```python
# filter onlt 1680 images
import os
import shutil

# Define your source and destination directories
src_folder = "data/source"
dest_folder = "data/etdrs"

# Create the destination folder if it doesn't already exist
os.makedirs(dest_folder, exist_ok=True)

# Supported image extensions (you can add more if needed)
image_extensions = (".jpg", ".jpeg", ".png", ".gif", ".bmp", ".tiff")

# Loop over all files in the source folder
for file_name in os.listdir(src_folder):
    # Check if '1680' is in the filename and the file is an image
    if "512" in file_name and file_name.lower().endswith(image_extensions):
        source_path = os.path.join(src_folder, file_name)
        dest_path = os.path.join(dest_folder, file_name)
        shutil.copy2(source_path, dest_path)
        print(f"Copied {file_name} to {dest_folder}")
```

```python
## output formatter
import json
import re
import tqdm
js_path = 'output/cogvlm.json'
file_name = js_path.split('/')[-1]
output = []
with open(js_path, 'r') as f:
    js1 = json.load(f)
    
for res in tqdm.tqdm(js1):
    rec_texts = res['rec_texts']
    text_li = []
    for text in re.finditer(r"'(.*?)'",rec_texts):
        text = text.group()
        text_li += re.sub(r"[^A-Za-z0-9 ]+","",text).split()
    # for text in re.sub(r"[^A-Za-z0-9 ]+","",rec_texts).split():
    for text in text_li:
        text = re.sub(r"unanswerable","",text)
        template = res.copy()
        template['rec_texts'] = text
        output.append(template)
with open(f'filtered/{file_name}', 'w') as f: 
    json.dump(output, f)
```

```python
import json
import re
import tqdm
import os
models = ['claude','git_b','git_l','mplug','qwen','gpt4o','llava','cogvlm','blip2_flan','GPT1','GPT2','pali']
for m in tqdm.tqdm(models):
    js_path = f'filtered/{m}.json'
    file_name = js_path.split('/')[-1][:-5]
    filter_dict = {}
    if os.path.exists(js_path):
        with open(js_path, 'r') as f:
            js1 = json.load(f)
            for item in js1:
                filter = item['filter']
                if filter not in filter_dict.keys():
                    filter_dict[filter] = []
                filter_dict[filter].append(item)
        for k,v in filter_dict.items():
            with open(f'/Users/kaiagao/Documents/Models/results/totaltext_new/{k}/{file_name}_totaltext.json', 'w') as f: 
                    json.dump(v, f)
```
